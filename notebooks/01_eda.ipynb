# Perfil descriptivo del universo de estudio

Objetivo: describir la clase multifatal por año sin convertir asociaciones en causalidad.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT=Path.cwd()
base=pd.read_parquet(ROOT/'data/processed/base_limpia.parquet')
base['year']=pd.to_datetime(base['FECHA']).dt.year
annual=base.groupby('year')['target_multifatal'].agg(['size','sum','mean'])
assert len(base)==9104 and set(annual.index)=={2021,2022,2023,2024,2025}
annual
```

## Lectura

La tasa es condicional a siniestros ya fatales. No representa riesgo de fatalidad entre todos los siniestros viales.

```python
ax=annual['mean'].mul(100).plot(kind='bar', color='#496B7C', figsize=(8,4), ylabel='Multifatalidad (%)', xlabel='Año')
ax.set_title('Tasa multifatal observada por año')
plt.tight_layout()
plt.show()
```

## Lectura de la figura

La comparación anual es descriptiva; 2025 es preliminar y no autoriza inferencias causales ni nuevos ajustes.

### Data Analysis Key Findings
- El universo contiene 9.104 siniestros fatales registrados.
- La clase positiva es FALLECIDOS >= 2.
